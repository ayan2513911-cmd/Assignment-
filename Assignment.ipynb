```python

import numpy as np

subjects = ["Maths", "English", "Science", "Computer", "Hindi"]

marks = np.random.randint(0, 101, size=(10, 5))

print("Subjects:", subjects)
print("Marks Matrix:")
print(marks)

total = np.sum(marks, axis=1)
percentage = total / 5

print("Total Marks:", total)
print("Percentage:", percentage)

avg = np.mean(marks, axis=0)
std = np.std(marks, axis=0)

print("Average Marks:", avg)
print("Standard Deviation:", std)

highest = np.max(marks, axis=0)
topper = np.argmax(marks, axis=0)

print("Highest Marks in Each Subject:", highest)
print("Topper Student Index:", topper)

passed = np.all(marks >= 35, axis=1)

print("Pass Students:", passed)
print("Pass Student Indices:", np.where(passed)[0])
print("Fail Student Indices:", np.where(~passed)[0])
```
